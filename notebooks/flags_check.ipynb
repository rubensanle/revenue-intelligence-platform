```python
import duckdb
con = duckdb.connect("../data/processed/revenue.duckdb", read_only=True)

# ¿Cuántos productos distintos tienen líneas > 10.000?
```

```python

df = con.execute("""
    SELECT
        stock_code,
        invoice_no,
        invoice_date,
        line_type,
        units_net_line,
        net_revenue_line
    FROM v_sales_lines
    WHERE is_product AND NOT is_partial_month
      AND line_type IN ('sale', 'cancellation', 'return')
      AND ABS(units_net_line) > 10000
    ORDER BY ABS(units_net_line) DESC
""").df()

print(df.to_string())
con.close()
```
